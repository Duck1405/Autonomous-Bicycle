```python
train_list = os.listdir(json_train)
print(f"Train list: {train_list[:5]}")

class_names = ["person", "vehicle"]
all_categories = {}

dir = os.listdir(json_train)

for i in range(len(dir)):
    with open(os.path.join(json_train, dir[i])) as f:
        lines = f.readlines()
    for line in lines:
        line = line.strip()
        if not line:
            continue
        class_id = int(line.split()[0])
        # print(class_id)
        category = class_names[class_id]
        if category in all_categories.keys():
            all_categories[category] += 1
        else:
            all_categories[category] = 1
        
    if i % 5000 == 0:
        print(f"Processed {i} files")
    

print(all_categories)
```
